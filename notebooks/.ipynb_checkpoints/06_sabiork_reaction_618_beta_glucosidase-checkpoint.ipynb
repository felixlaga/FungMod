```python
from pathlib import Path
import shutil
import yaml
import json

from fungal_model.registry import load_registry
from fungal_model.screening import assess_modelability, simulate_screen

# ---------------------------------------------------------------------
# 1. User-chosen exploratory range for the missing enzyme concentration
# ---------------------------------------------------------------------
#
# Current missing parameter:
# enzyme_concentration_beta_glucosidase
#
# Units are mM because the existing unknown ValueSpec in your report uses mM.
# This is NOT literature-curated. It is an exploratory assumption.
#
# Range below:
# 1e-6 mM to 1e-3 mM
# = 1 nM to 1 µM
#
# You can change these bounds later.
ENZYME_CONC_LOWER_M_MOLAR = 1.0e-6
ENZYME_CONC_UPPER_M_MOLAR = 1.0e-3

N_SAMPLES = 32
SEED = 1

# ---------------------------------------------------------------------
# 2. Make a temporary exploratory registry copy
# ---------------------------------------------------------------------

base_registry_dir = ROOT / "data_registry"
exploratory_registry_dir = ROOT / "outputs" / "tmp_sabiork_reaction_618_exploratory_registry"

if exploratory_registry_dir.exists():
    shutil.rmtree(exploratory_registry_dir)

shutil.copytree(base_registry_dir, exploratory_registry_dir)

parameter_file = exploratory_registry_dir / "parameters" / "parameter_records.yml"

if not parameter_file.exists():
    raise FileNotFoundError(f"Could not find parameter registry file: {parameter_file}")

with parameter_file.open("r", encoding="utf-8") as f:
    parameter_data = yaml.safe_load(f)

# Support both possible YAML shapes:
#   - a plain list of records
#   - {"records": [...]}
if isinstance(parameter_data, list):
    records = parameter_data
elif isinstance(parameter_data, dict) and isinstance(parameter_data.get("records"), list):
    records = parameter_data["records"]
else:
    raise TypeError(
        "Unsupported parameter_records.yml shape. Expected a list or a dict with key 'records'."
    )

target_symbol = "enzyme_concentration_beta_glucosidase"
target_record_id = "sabiork_reaction_618_enzyme_concentration_beta_glucosidase"

target_records = [
    r for r in records
    if r.get("parameter_symbol") == target_symbol or r.get("record_id") == target_record_id
]

if not target_records:
    raise KeyError(
        f"Could not find parameter record for {target_symbol}. "
        "Check the parameter_records.yml file and the record ID."
    )

if len(target_records) > 1:
    print(f"Warning: found {len(target_records)} matching records. Updating all of them.")

for record in target_records:
    record["maturity"] = "exploratory_prior"
    record["value"] = {
        "kind": "distribution",
        "distribution": "loguniform",
        "parameters": {
            "lower": ENZYME_CONC_LOWER_M_MOLAR,
            "upper": ENZYME_CONC_UPPER_M_MOLAR,
        },
        "units": "mM",
        "source": "user-supplied exploratory range in notebook",
        "confidence_level": "exploratory_assumption",
        "notes": (
            "SABIO-RK EntryID 35622 does not provide enzyme concentration. "
            "This range is not literature-curated and must not be used as a scientific claim."
        ),
    }
    record.setdefault("provenance", {})
    record["provenance"]["exploratory_override"] = True
    record["provenance"]["original_reason"] = (
        "enzyme_concentration_beta_glucosidase was unknown in the selected SABIO-RK snapshot."
    )
    record["notes"] = (
        "Exploratory notebook override for missing enzyme concentration. "
        "Original SABIO-RK-curated record remains unknown in the base registry."
    )

with parameter_file.open("w", encoding="utf-8") as f:
    yaml.safe_dump(parameter_data, f, sort_keys=False)

print(f"Wrote exploratory registry copy to: {exploratory_registry_dir}")
print(f"Updated parameter: {target_symbol}")
print(
    "Exploratory enzyme concentration range:",
    ENZYME_CONC_LOWER_M_MOLAR,
    "to",
    ENZYME_CONC_UPPER_M_MOLAR,
    "mM",
)

# ---------------------------------------------------------------------
# 3. Reload registry and check modelability in exploratory mode
# ---------------------------------------------------------------------

exploratory_registry = load_registry(exploratory_registry_dir / "registry_index.yml")

exploratory_report = assess_modelability(
    fungus_id=FUNGUS_ID,
    substrate_id=SUBSTRATE_ID,
    environment_id=ENVIRONMENT_ID,
    registry=exploratory_registry,
    mode="exploratory",
)

print(exploratory_report.summary())
exploratory_report.to_dict()
```

```python
from pathlib import Path
import shutil
import yaml
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from fungal_model.registry import load_registry
from fungal_model.screening import assess_modelability, build_model_config_from_registry_case
from fungal_model.workflows import run_configured_model

# ---------------------------------------------------------------------
# Manual exploratory ensemble for homogeneous_michaelis_menten
# because simulate_screen currently only supports surface_catalysis.
# ---------------------------------------------------------------------

ENZYME_CONC_LOWER_M_MOLAR = 1.0e-6   # mM = 1 nM
ENZYME_CONC_UPPER_M_MOLAR = 1.0e-3   # mM = 1 µM

N_SAMPLES = 32
SEED = 1

base_registry_dir = ROOT / "data_registry"
manual_output_dir = ROOT / "outputs" / "sabiork_reaction_618_manual_homogeneous_mm_ensemble"
sample_registry_root = manual_output_dir / "sample_registries"
sample_run_root = manual_output_dir / "sample_runs"

if manual_output_dir.exists():
    shutil.rmtree(manual_output_dir)

sample_registry_root.mkdir(parents=True, exist_ok=True)
sample_run_root.mkdir(parents=True, exist_ok=True)

rng = np.random.default_rng(SEED)

# loguniform sampling between lower and upper
enzyme_concentrations = np.exp(
    rng.uniform(
        np.log(ENZYME_CONC_LOWER_M_MOLAR),
        np.log(ENZYME_CONC_UPPER_M_MOLAR),
        size=N_SAMPLES,
    )
)

def update_enzyme_concentration_to_exact(parameter_file: Path, value_mM: float) -> None:
    """Change the enzyme concentration parameter to an exact sampled value."""
    with parameter_file.open("r", encoding="utf-8") as f:
        parameter_data = yaml.safe_load(f)

    if isinstance(parameter_data, list):
        records = parameter_data
    elif isinstance(parameter_data, dict) and isinstance(parameter_data.get("records"), list):
        records = parameter_data["records"]
    else:
        raise TypeError(
            "Unsupported parameter_records.yml shape. Expected list or {'records': [...]}."
        )

    target_symbol = "enzyme_concentration_beta_glucosidase"
    target_record_id = "sabiork_reaction_618_enzyme_concentration_beta_glucosidase"

    matches = [
        r for r in records
        if r.get("parameter_symbol") == target_symbol or r.get("record_id") == target_record_id
    ]

    if not matches:
        raise KeyError(f"Could not find parameter record for {target_symbol}")

    for record in matches:
        record["maturity"] = "exploratory_sample"
        record["value"] = {
            "kind": "exact",
            "value": float(value_mM),
            "units": "mM",
            "source": "sampled from user-supplied exploratory loguniform range in notebook",
            "confidence_level": "exploratory_assumption",
            "notes": (
                "Sampled enzyme concentration. This is not SABIO-RK-curated and "
                "must not be treated as a literature value."
            ),
        }

        record.setdefault("provenance", {})
        record["provenance"]["exploratory_sample"] = True
        record["provenance"]["sampled_from_range_mM"] = {
            "lower": ENZYME_CONC_LOWER_M_MOLAR,
            "upper": ENZYME_CONC_UPPER_M_MOLAR,
            "distribution": "loguniform",
        }

    with parameter_file.open("w", encoding="utf-8") as f:
        yaml.safe_dump(parameter_data, f, sort_keys=False)


def extract_final_states(result):
    """
    Try to extract final state values from a SimulationResult robustly.
    This is deliberately defensive because the exact result object shape may evolve.
    """
    if hasattr(result, "to_dict"):
        d = result.to_dict()
    elif isinstance(result, dict):
        d = result
    else:
        d = result.__dict__

    # Common possible structures
    candidates = [
        d.get("final_states") if isinstance(d, dict) else None,
        d.get("final_state") if isinstance(d, dict) else None,
        d.get("states") if isinstance(d, dict) else None,
        d.get("state_trajectories") if isinstance(d, dict) else None,
    ]

    for candidate in candidates:
        if isinstance(candidate, dict):
            # If already final states, return directly.
            if all(not isinstance(v, (list, tuple)) for v in candidate.values()):
                return candidate

            # If trajectories, take final point.
            final = {}
            for key, value in candidate.items():
                if isinstance(value, dict) and "values" in value:
                    values = value["values"]
                    final[key] = values[-1] if values else None
                elif isinstance(value, (list, tuple)) and len(value) > 0:
                    final[key] = value[-1]
                else:
                    final[key] = value
            return final

    return {}


rows = []
failures = []

for i, enzyme_conc_mM in enumerate(enzyme_concentrations):
    sample_id = f"sample_{i:03d}"

    sample_registry_dir = sample_registry_root / sample_id
    sample_output_dir = sample_run_root / sample_id

    shutil.copytree(base_registry_dir, sample_registry_dir)

    parameter_file = sample_registry_dir / "parameters" / "parameter_records.yml"
    update_enzyme_concentration_to_exact(parameter_file, enzyme_conc_mM)

    sample_registry = load_registry(sample_registry_dir / "registry_index.yml")

    sample_report = assess_modelability(
        fungus_id=FUNGUS_ID,
        substrate_id=SUBSTRATE_ID,
        environment_id=ENVIRONMENT_ID,
        registry=sample_registry,
        mode="scientific",
    )

    if sample_report.status != "modelable":
        failures.append({
            "sample": i,
            "enzyme_concentration_mM": float(enzyme_conc_mM),
            "status": sample_report.status,
            "report": sample_report.to_dict(),
        })
        continue

    try:
        config = build_model_config_from_registry_case(
            fungus_id=FUNGUS_ID,
            substrate_id=SUBSTRATE_ID,
            environment_id=ENVIRONMENT_ID,
            registry=sample_registry,
            mode="scientific",
            output_directory=str(sample_output_dir),
        )

        sample_output_dir.mkdir(parents=True, exist_ok=True)
        config_path = sample_output_dir / "model_config.yml"
        config_path.write_text(
            yaml.safe_dump(config.to_dict(), sort_keys=False),
            encoding="utf-8",
        )

        result = run_configured_model(
            config_path,
            output_dir=sample_output_dir / "bundle",
        )

        # Save plot for a few samples only, otherwise notebook/output gets cluttered.
        if i < 5:
            try:
                result.plot_states(sample_output_dir / "states.png")
            except Exception as plot_error:
                print(f"Plot failed for {sample_id}: {plot_error}")

        final_states = extract_final_states(result)

        row = {
            "sample": i,
            "enzyme_concentration_mM": float(enzyme_conc_mM),
            "status": sample_report.status,
        }

        for key, value in final_states.items():
            if isinstance(value, dict):
                row[f"final_{key}"] = value.get("value", value.get("magnitude", value))
            else:
                row[f"final_{key}"] = value

        rows.append(row)

    except Exception as exc:
        failures.append({
            "sample": i,
            "enzyme_concentration_mM": float(enzyme_conc_mM),
            "error": repr(exc),
        })

df = pd.DataFrame(rows)
failures_df = pd.DataFrame(failures)

manual_output_dir.mkdir(parents=True, exist_ok=True)
df.to_csv(manual_output_dir / "manual_ensemble_summary.csv", index=False)
failures_df.to_csv(manual_output_dir / "manual_ensemble_failures.csv", index=False)

metadata = {
    "kind": "manual_homogeneous_michaelis_menten_exploratory_ensemble",
    "reason": (
        "simulate_screen currently supports only surface_catalysis; this notebook "
        "manually samples enzyme_concentration_beta_glucosidase and runs deterministic "
        "homogeneous_michaelis_menten configs."
    ),
    "n_samples_requested": N_SAMPLES,
    "n_samples_succeeded": len(df),
    "n_samples_failed": len(failures_df),
    "enzyme_concentration_range_mM": {
        "lower": ENZYME_CONC_LOWER_M_MOLAR,
        "upper": ENZYME_CONC_UPPER_M_MOLAR,
        "distribution": "loguniform",
    },
    "warning": (
        "The enzyme concentration range is user-supplied exploratory input, "
        "not a SABIO-RK literature value."
    ),
}

(manual_output_dir / "manual_ensemble_metadata.json").write_text(
    json.dumps(metadata, indent=2),
    encoding="utf-8",
)

print("Manual exploratory ensemble complete.")
print(f"Succeeded: {len(df)} / {N_SAMPLES}")
print(f"Failed: {len(failures_df)} / {N_SAMPLES}")
print(f"Output directory: {manual_output_dir}")

display(df.head())
if not failures_df.empty:
    display(failures_df.head())
```

```python
import ast
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

%matplotlib inline

manual_output_dir = ROOT / "outputs" / "sabiork_reaction_618_manual_homogeneous_mm_ensemble"
summary_path = manual_output_dir / "manual_ensemble_summary.csv"

df = pd.read_csv(summary_path)

display(df.head())
display(df.describe(include="all"))

# ------------------------------------------------------------
# Helper: many final-state columns were saved as lists in strings.
# We convert "[1, 2, 3]" into the last value of the trajectory.
# ------------------------------------------------------------

def last_value(x):
    if isinstance(x, (int, float)):
        return x

    if pd.isna(x):
        return None

    if isinstance(x, str):
        x = x.strip()

        # Try parsing stringified lists like "[3.06, 3.05, ...]"
        if x.startswith("[") and x.endswith("]"):
            try:
                parsed = ast.literal_eval(x)
                if isinstance(parsed, list) and len(parsed) > 0:
                    return parsed[-1]
            except Exception:
                return None

        # Try parsing normal numeric strings
        try:
            return float(x)
        except Exception:
            return None

    return None


# ------------------------------------------------------------
# Build a cleaned dataframe with scalar final values.
# ------------------------------------------------------------

clean = df.copy()

for col in clean.columns:
    if col.startswith("final_"):
        clean[col + "_last"] = clean[col].apply(last_value)

display(clean.head())

clean.to_csv(manual_output_dir / "manual_ensemble_summary_cleaned.csv", index=False)

# ------------------------------------------------------------
# Plot sampled enzyme concentration.
# ------------------------------------------------------------

plt.figure()
plt.hist(clean["enzyme_concentration_mM"], bins=12)
plt.xscale("log")
plt.xlabel("enzyme_concentration_beta_glucosidase (mM)")
plt.ylabel("count")
plt.title("Sampled enzyme concentration")
plt.show()

# ------------------------------------------------------------
# Plot final values of each trajectory state.
# ------------------------------------------------------------

last_cols = [c for c in clean.columns if c.startswith("final_") and c.endswith("_last")]

for col in last_cols:
    numeric = pd.to_numeric(clean[col], errors="coerce").dropna()

    if len(numeric) == 0:
        print(f"Skipping {col}: no numeric values found.")
        continue

    plt.figure()
    plt.hist(numeric, bins=12)
    plt.xlabel(col)
    plt.ylabel("count")
    plt.title(f"Final-state distribution: {col}")
    plt.show()

# ------------------------------------------------------------
# Optional: relationship between sampled enzyme concentration
# and final product concentration.
# ------------------------------------------------------------

product_candidates = [
    c for c in last_cols
    if "glucose" in c.lower() or "product" in c.lower()
]

for product_col in product_candidates:
    product_values = pd.to_numeric(clean[product_col], errors="coerce")

    plt.figure()
    plt.scatter(clean["enzyme_concentration_mM"], product_values)
    plt.xscale("log")
    plt.xlabel("enzyme_concentration_beta_glucosidase (mM)")
    plt.ylabel(product_col)
    plt.title(f"Product after simulation vs enzyme concentration")
    plt.show()

clean
```

## Limitations

- no biomass growth;
