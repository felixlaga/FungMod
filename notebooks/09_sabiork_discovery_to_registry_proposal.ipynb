# SOURCE-002 SABIO-RK Discovery to Registry Proposal

This notebook demonstrates a review-only SABIO-RK discovery workflow. It uses frozen local snapshots by default, generates proposed FungMod records, and writes proposal files that are not production registry records.

Every generated record is review required. Nothing in this notebook promotes proposals into the simulation registry.

```python
import os
from pathlib import Path

from fungal_model.sources.sabiork import SabioRKSource

output_root = Path(os.environ.get("FUNGMOD_NOTEBOOK_OUTPUT_ROOT", "outputs/notebooks"))
proposal_output_dir = output_root / "09_sabiork_discovery_to_registry_proposal"

source = SabioRKSource(cache_dir="data/source_snapshots/sabiork")
```

Discover entries by researcher-facing names and identifiers. The local repository contains a frozen Reaction 618 snapshot, so this path does not call live SABIO-RK.

```python
discovery = source.discover_for_virtual_experiment(
    fungus="Oryza sativa",
    substrate="cellobiose",
    enzyme="beta-glucosidase",
    ec_number="3.2.1.21",
    reaction_id="618",
    entry_id="35622",
)

reaction_rows = discovery.show_reactions()
product_rows = discovery.show_products()
parameter_rows = discovery.show_kinetic_parameters()
missing_field_rows = discovery.show_missing_fields()
```

Create review-only proposed records. The output bundle is suitable for human curation before any separate promotion step.

```python
proposal = discovery.to_registry_proposal(
    process_type="homogeneous_michaelis_menten",
    product_map="auto_from_stoichiometry",
)

proposal_preview = proposal.preview()
write_result = proposal.write(proposal_output_dir)
written_files = write_result.to_dict()
```
