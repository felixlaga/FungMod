# Generic Surface Hydrolysis Demo

Use the same generic surface-catalysis components for two toy solid substrates.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from fungal_model import AccessibleSitePool, AccessibleSurfaceAreaModel, LangmuirAdsorptionModel, ProductReleaseMap, SurfaceCatalysisModel, SurfaceCatalysisProcess
from fungal_model.core.units import Q_
```

```python
shared_adsorption = LangmuirAdsorptionModel("K_ads", "mole / liter", "Toy surface demo adsorption model.")
shared_catalysis = SurfaceCatalysisModel("k_surface", "kilogram / second", "Toy surface demo catalysis model.")

cellulose_process = SurfaceCatalysisProcess(
    name="toy cellulose surface cleavage",
    substrate_state="cellulose_mass",
    enzyme_state="E",
    substrate_units="kilogram",
    enzyme_units="mole / liter",
    accessible_site_pool=AccessibleSitePool("cellulose accessible surface", "beta-1,4-glycosidic"),
    accessible_surface_model=AccessibleSurfaceAreaModel.from_quantity(name="cellulose area", area=Q_(0.5, "meter ** 2")),
    adsorption_model=shared_adsorption,
    catalytic_model=shared_catalysis,
    product_release_map=ProductReleaseMap.one_to_one(substrate_state="cellulose_mass", product_state="sugar_equivalent"),
)

chitin_process = SurfaceCatalysisProcess(
    name="toy chitin surface cleavage",
    substrate_state="chitin_mass",
    enzyme_state="E",
    substrate_units="kilogram",
    enzyme_units="mole / liter",
    accessible_site_pool=AccessibleSitePool("chitin accessible surface", "beta-1,4-glycosidic"),
    accessible_surface_model=AccessibleSurfaceAreaModel.from_quantity(name="chitin area", area=Q_(0.25, "meter ** 2")),
    adsorption_model=shared_adsorption,
    catalytic_model=shared_catalysis,
    product_release_map=ProductReleaseMap.one_to_one(substrate_state="chitin_mass", product_state="oligomer_equivalent"),
)

[cellulose_process.to_dict(), chitin_process.to_dict()]
```
