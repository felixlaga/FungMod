# Config And Entity Inspection

Load a non-plugin surface benchmark config and inspect the config references and saved entity snapshots.

```python
from pathlib import Path
import json
import sys

ROOT = Path.cwd()
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from fungal_model import load_model_config, run_configured_model

CONFIG = ROOT / "data" / "model_configs" / "toy_surface_dummy_non_pet.yml"
OUTPUT = ROOT / "notebooks" / "examples" / "Outputs" / "01_config_entity_inspection"
```

```python
config = load_model_config(CONFIG)
config_overview = {
    "name": config.name,
    "entities": config.entities.to_dict(),
    "processes": [process.to_dict() for process in config.processes],
}
config_overview
```

```python
run_configured_model(CONFIG, output_dir=OUTPUT)
entity_index = json.loads((OUTPUT / "entity_snapshots" / "index.json").read_text(encoding="utf-8"))
sorted({entry["role"] for entry in entity_index["entities"]})
```
