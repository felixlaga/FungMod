# Structured Failure Report (software-test fixture only)

This notebook uses a toy plugin config fixture to exercise failure reporting. It is not a researcher-facing scientific example and its outputs must not be cited as biological evidence.

Show that unsupported configured execution fails with a structured report instead of a hidden fallback.


```python
from pathlib import Path
import json
import os
import sys

ROOT = Path.cwd()
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from fungal_model import ConfiguredModelExecutionError, run_configured_model

CONFIG = ROOT / "data" / "model_configs" / "toy_surface_pet_plugin.yml"
OUTPUT_ROOT = Path(os.environ.get("FUNGMOD_NOTEBOOK_OUTPUT_ROOT", str(ROOT / "notebooks" / "examples" / "Outputs")))
OUTPUT = OUTPUT_ROOT / "02_failure_report"
```

```python
report = None
try:
    run_configured_model(CONFIG, output_dir=OUTPUT)
except ConfiguredModelExecutionError as exc:
    report = exc.report.to_dict()

if report is None:
    raise RuntimeError("expected the plugin config to require an explicit registry")

OUTPUT.mkdir(parents=True, exist_ok=True)
(OUTPUT / "failure_report.json").write_text(json.dumps(report, indent=2, sort_keys=True) + "\n", encoding="utf-8")
report
```
