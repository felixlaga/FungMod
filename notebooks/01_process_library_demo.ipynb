# Process Library Demo

Inspect registered processes, their required inputs, matching behavior, and structured missing-process reports.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from fungal_model import FirstOrderDecayProcess, ModelBuilder, ProcessRegistry
from fungal_model.core.errors import MissingProcessError
from fungal_model.core.parameters import Parameter, ParameterSet
```

```python
process = FirstOrderDecayProcess(
    name="toy first-order decay",
    substrate_state="S",
    product_state="P",
    rate_constant_symbol="k",
    state_units="mole / liter",
)
registry = ProcessRegistry([process])
registry.to_dict()
```

```python
parameters = ParameterSet([
    Parameter("toy first-order rate", "k", 0.1, "1 / second", 0.0, "Toy process-library value.", "testing", "Notebook demo."),
])
assembled = ModelBuilder(
    process_library=registry,
    requested_processes=("first_order_decay",),
    parameters=parameters,
).assemble()
assembled.assembly_report.to_dict()
```

```python
try:
    ModelBuilder(process_library=registry, requested_processes=("surface_catalysis",), parameters=parameters).assemble()
except MissingProcessError as exc:
    missing_report = exc.report.to_dict()
missing_report
```
